```python
import random

class TicTacToe:
    def __init__(self):
        """Initialize empty 3x3 board and set starting player."""
        self.board = [" "] * 9
        self.current_player = "X"

    def display_board(self):
        """Render current grid state to terminal."""
        print("\n")
        for i in range(0, 9, 3):
            print(f" {self.board[i]} | {self.board[i+1]} | {self.board[i+2]} ")
            if i < 6:
                print("---|---|---")
        print("\n")

    def is_valid_move(self, position):
        """Check if index is within bounds and target cell is empty."""
        return 0 <= position <= 8 and self.board[position] == " "

    def make_move(self, position):
        """Place active player's symbol in selected index."""
        if self.is_valid_move(position):
            self.board[position] = self.current_player
            return True
        return False

    def check_winner(self):
        """Evaluate 8 winning combinations or draw state."""
        win_conditions = [
            (0, 1, 2), (3, 4, 5), (6, 7, 8),  # Rows
            (0, 3, 6), (1, 4, 7), (2, 5, 8),  # Columns
            (0, 4, 8), (2, 4, 6)             # Diagonals
        ]
        for a, b, c in win_conditions:
            if self.board[a] == self.board[b] == self.board[c] != " ":
                return self.board[a]
        if " " not in self.board:
            return "Draw"
        return None

    def switch_player(self):
        """Toggle active turn between X and O."""
        self.current_player = "O" if self.current_player == "X" else "X"


def get_computer_move(game):
    """Select a random free cell for computer move."""
    empty_positions = [i for i, cell in enumerate(game.board) if cell == " "]
    return random.choice(empty_positions)


def main():
    """Main execution loop for Tic Tac Toe application."""
    print("=================================")
    print("   WELCOME TO TIC TAC TOE GAME   ")
    print("=================================")

    while True:
        print("\nSelect Game Mode:")
        print("1. Player vs Player")
        print("2. Player vs Computer")
        mode = input("Enter choice (1 or 2): ").strip()
        
        if mode in ["1", "2"]:
            break
        print("Invalid choice. Please enter 1 or 2.")

    while True:
        game = TicTacToe()
        
        while True:
            game.display_board()
            print(f"Player {game.current_player}'s turn.")

            # Get move index
            if mode == "2" and game.current_player == "O":
                position = get_computer_move(game)
                print(f"Computer chose position: {position + 1}")
            else:
                try:
                    move = int(input("Enter position (1-9): ")) - 1
                    if not game.is_valid_move(move):
                        print("Invalid move. Cell occupied or out of bounds (1-9). Try again.")
                        continue
                    position = move
                except ValueError:
                    print("Invalid input. Please enter a valid number (1-9).")
                    continue

            game.make_move(position)
            result = game.check_winner()

            # End game if winner or draw is found
            if result:
                game.display_board()
                if result == "Draw":
                    print("Game Over! It's a draw!")
                else:
                    print(f"Game Over! Player {result} wins!")
                break

            game.switch_player()

        replay = input("\nDo you want to play again? (y/n): ").strip().lower()
        if replay != "y":
            print("Thanks for playing! Goodbye.")
            break


if __name__ == "__main__":
    main()
```
